```python
import os
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Step 0: Create preview directory if it doesn't exist
os.makedirs("preview", exist_ok=True)

# Step 1: Data augmentation setup
# this augumentation also not all types are applied to the one image only one type is applied to that image and it is in random way
datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=40,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)

# Step 2: Load images from directory
# here images are selected randomly from the parent folder and augumeneted that image
train_generator = datagen.flow_from_directory(
    directory="PetImages", # parent folder 
    target_size=(150, 150), # resize images to this 
    batch_size=1, # Loads 1 image per batch
    class_mode='binary', # labels assigned to cat and dog as 0 and 1
    save_to_dir="preview",  # ✔ Saves augmented images to the preview folder
    save_prefix="aug", # ✔ All saved images start with aug_
    save_format="jpeg" 
)

# Step 3: Generate and save 20 augmented images
i = 0
for _ in train_generator:
    i += 1
    if i >= 20:
        break
```

* Image Generator: it is used for image augumentation, and loading images from folders and Feeding images to deep learning models
* rescale=1./255 : Converts pixel values from 0–255 → 0–1 for train faster
* rotation_range=40 : ✔ Randomly rotates images between –40° to +40°
* width_shift_range=0.2 : ✔ Randomly shifts image horizontally by 20%
* height_shift_range=0.2 : ✔ Randomly shifts image vertically by 20%
* shear_range=0.2 : Slants the image slightly
* zoom_range=0.2 : Randomly zooms in or out by 20%
* horizontal_flip=True : ✔ Randomly flips image left ↔ right
* fill_mode='nearest' : ✔ Fills empty pixels created after transformation and ✔ Uses nearest pixel value to fill gaps
